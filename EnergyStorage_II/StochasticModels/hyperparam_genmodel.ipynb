##  Hyperparamtuning of the stochastic model
The model is a Conditional Normalizing Flow (CNF) with a Diagonal Gaussian base distribution and Masked Autoregressive Flow (MAF) layers as flows.
Using only lag_1 and lag_2 of target as context. Objective for the optuna trial is the wasserstein distance metric.

### Preprocessing data and creating features

```python
# Load train and validation data

genmodel_df = pd.read_csv("../Data/genmodel.csv")
policy_df = pd.read_csv("../Data/policy.csv")
```

```python
genmodel_df["mean_markets"] = genmodel_df[["DAP FR", "DAP DK1", "DAP CH", "DAP BE", "DAP BE"]].mean(axis=1)
policy_df["mean_markets"] = policy_df[["DAP FR", "DAP DK1", "DAP CH", "DAP BE", "DAP BE"]].mean(axis=1)

columns_to_shift = genmodel_df.columns.difference(["DAP DE/LU", "Residual load forecast", "Total Consumption forecast", "Total production forecast"])
genmodel_df[columns_to_shift] = genmodel_df[columns_to_shift].shift(1)
policy_df[columns_to_shift] = policy_df[columns_to_shift].shift(1)
```

```python
def get_1_day_lag(df):
    df_copy = df.copy()
    for lag in range(1, 25):
        df_copy['lag_{}'.format(lag)] = df_copy["DAP DE/LU"].shift(lag)
    df_copy = df_copy.dropna()
    df_copy = df_copy.reset_index(drop=True)    
    return df_copy

genmodel_df = get_1_day_lag(genmodel_df)
policy_df = get_1_day_lag(policy_df)
```

### Define plotting functions and metrics

```python
import plotly.graph_objects as go
from scipy.stats import ks_2samp
from scipy.stats import wasserstein_distance
from StochasticModels.Plotting import plot_samples, plot_time_series_with_bounds

def plot_targets_vs_pred(targets, predictions):
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=predictions, y=targets, mode="markers", name="Targets"))
    fig.add_trace(go.Scatter(x=[min(targets), max(targets)], y=[min(targets), max(targets)], mode="lines", name="Predictions"))
    fig.update_layout(title="Targets vs Predictions", xaxis_title="Predictions", yaxis_title="Targets")
    fig.show()

def plot_predictions_over_t(targets, predictions):
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=[i for i in range(len(targets))], y=targets, mode="lines", name="Targets"))
    fig.add_trace(go.Scatter(x=[i for i in range(len(predictions))], y=predictions, mode="lines", name="Predictions"))
    fig.update_layout(title="Targets vs Predictions", xaxis_title="Time", yaxis_title="Price")
    fig.show()

def calc_avg_mse(rescaled_sample_tensor, X_val_rescaled_tensor):
    mse = nn.MSELoss()
    mse_scores = [mse(rescaled_sample_tensor[i, :, 0], X_val_rescaled_tensor) for i in range(rescaled_sample_tensor.shape[0])]
    return np.mean(mse_scores)

def calc_sdch_sdcv(rescaled_samples_tensor):
    H = rescaled_samples_tensor.shape[1]
    N = rescaled_samples_tensor.shape[0]

    SDCH_sum = 0
    SDCV_sum = 0
    for t in range(H-1):
        diff = torch.abs(rescaled_samples_tensor[:, t+1, 0] - rescaled_samples_tensor[:, t, 0])
        SDCH_sum += diff.mean()
    SDCH = SDCH_sum /  (H-1)

    for i in range(N):
        diff = torch.abs(rescaled_samples_tensor[i, :, 0] - rescaled_samples_tensor[i-1, :, 0])
        SDCV_sum += diff.mean()
    SDCV = SDCV_sum / (H-1)

    return SDCH, SDCV   

def ks_score(rescaled_samples, X_test_rescaled):
    ks_scores = []
    for i in range(rescaled_samples.shape[0]):
        sample = rescaled_samples[i, :, 0].numpy()
        X_test = X_test_rescaled.numpy()
        ks_test = ks_2samp(sample, X_test)
        ks_scores.append(ks_test[0] if ks_test[1] < 0.05 else 1.0)
    return np.mean(ks_scores)

def wd_score(rescaled_samples, X_test_rescaled):
    """
    Function to calculate the Wasserstein dsitance between the generated samples and the avalidation data.
    """
    wd_scores = []  
    for i in range(rescaled_samples.shape[0]):
        sample = rescaled_samples[i, :, 0].numpy()
        X_test = X_test_rescaled.numpy()
        wd_scores.append(wasserstein_distance(sample, X_test))
        
    return np.mean(wd_scores)
```

### Optuna objective and run study functions

```python
import optuna
from StochasticModels.CNFModels import CNF

def objective_sampling(trial, X_train, c_train, X_test, c_test, X_scaler):
    """
    Objective function for the Optuna hyperparameter optimization.
    Trains a CNF model on the given data and returns the validation score. (combined score of Wasserstein (WD - for sensitivity) and Kolmogorov-Smirnov (KS - for accuracy) distance)
    Model is trained without a trainloader to ensure that the model is trained on the full dataset if training breaks because of negative kl-divergence values. 
    
    :Params:
    -------
    :trial: optuna.trial.Trial object - The trial object that is used to sample hyperparameters
    :X_train: torch.Tensor - The training data for the target variable
    :c_train: torch.Tensor - The training data for the conditional variables
    :X_test: torch.Tensor - The validation data for the target variable
    :c_test: torch.Tensor - The validation data for the conditional variables
    :X_scaler: sklearn.preprocessing.StandardScaler - The scaler object that was used to scale the target variable
    """

    # Hyperparameters for CNF model
    trainable_q0 = trial.suggest_categorical("trainable_q0", [True, False])
    num_flows = trial.suggest_int("num_flows", 1, 10)
    weight_decay = trial.suggest_categorical("weight_decay", [1e-5, 1e-4, 1e-3])
    temperature = trial.suggest_categorical("temperature", [0.5, 1.0, 1.25, 1.5, 1.75, 2.0])
    dropout_probability = trial.suggest_categorical("dropout_probability", [0.0, 0.1, 0.15, 0.2, 0.25, 0.3])
    hidden_features = trial.suggest_categorical("hidden_features", [32, 64, 128, 256])
    use_residuals = trial.suggest_categorical("use_residuals", [True, False])
    num_blocks = trial.suggest_int("num_blocks", 1, 5) if use_residuals else 0
    use_batch_norm = trial.suggest_categorical("use_batch_norm", [True, False])
    lr = trial.suggest_categorical("lr", [1e-4, 1e-3, 1e-2])

    # Initialize the CNF model
    model = CNF(
        x_dim=X_train.shape[1],
        c_dim=c_train.shape[1],
        trainable_q0=trainable_q0,
        num_flows=num_flows,
        hidden_features=hidden_features,
        num_blocks=num_blocks,
        dropout_probability=dropout_probability,
        temperature=temperature,
        weight_decay=weight_decay,
        use_residual_blocks=use_residuals,
        use_batch_norm=use_batch_norm,
        lr=lr
    )
    # Train the model on the training data
    model.fit(X_train, c_train, num_epochs=1000, relative_improvement_threshold=1e-2, verbose=False)
    
    # Generate samples from the trained model to calculate metrics
    num_samples = len(X_test)
    num_sample_paths = 100
    samples = model.generate_sample_paths(num_samples=num_samples, context=c_test, num_sample_paths=num_sample_paths)
    rescaled_samples = [X_scaler.inverse_transform(sample) for sample in samples]
    X_test_rescaled = X_scaler.inverse_transform(X_test).flatten()

    # Calculate the wasserstein distance between the generated samples and the validation data
    rescaled_samples_tensor = torch.tensor(np.array(rescaled_samples), dtype=torch.float32)
    X_test_rescaled_tensor = torch.tensor(X_test_rescaled, dtype=torch.float32)
    wd = wd_score(rescaled_samples_tensor, X_test_rescaled_tensor)
    sdch, sdcv = calc_sdch_sdcv(rescaled_samples_tensor)
    avg_mse = calc_avg_mse(rescaled_samples_tensor, X_test_rescaled_tensor)

    score = wd + sdch + sdcv + 0.1 * avg_mse

    if trial.should_prune():
        raise optuna.TrialPruned()

    return score

def run_study(obj_direction, objective_fn, study_name, X_train, c_train, X_test, c_test, X_scaler, n_trials=100):
    """
    Function to run a hyperparameter optimization study with Optuna.
    """
    study = optuna.create_study(direction=obj_direction, study_name=study_name)
    study.optimize(lambda trial: objective_fn(trial, X_train, c_train, X_test, c_test, X_scaler), n_trials=n_trials)
    print(f"Best trial: {study.best_trial.value}")
    print(f"Best params: {study.best_trial.params}")
    return study

def return_sample_target_df(sample_paths, target_df):
    sample_paths_array = np.array(sample_paths)
    sample_paths_df = sample_paths_array.reshape(sample_paths_array.shape[0], -1).T
    df_samples = pd.DataFrame(sample_paths_df, columns=[f"Scenario_{i+1}" for i in range(sample_paths_df.shape[1])]).reset_index(drop=True)
    merged_df = target_df.merge(df_samples, left_index=True, right_index=True)
    return merged_df
```

```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np

def multi_1d_dist(df, scen_dfs):
    fig = go.Figure()
    colors = px.colors.qualitative.Set1
    # overlap_color = "rgba(150, 0, 0, 0.05)"
    overlap_color = "rgba(0, 0, 255, 0.05)"
    
    global_min, global_max = None, None  

    for i, scen_df in enumerate(scen_dfs):
        min_values = scen_df.min(axis=1)
        max_values = scen_df.max(axis=1)
        color = colors[1]
        #color = colors[i % len(colors)]  

        if global_min is None:
            global_min = min_values
            global_max = max_values
        else:
            global_min = np.maximum(global_min, min_values)  
            global_max = np.minimum(global_max, max_values)  

        fig.add_trace(go.Scatter(
            x=scen_df.index.values[:24], y=min_values,
            mode="lines", name=f"Szenarien",
            line=dict(color=color),
            showlegend=True  
        ))
        fig.add_trace(go.Scatter(
            x=scen_df.index.values[:24], y=max_values,
            mode="lines", name=f"Max Scenario {i+1}",
            line=dict(color=color),
            showlegend=False,  
            fill="tonexty"
        ))

    fig.add_trace(go.Scatter(
        x=scen_dfs[0].index.values[:24], y=global_min,
        mode="lines", name="Overlap Min",
        line=dict(color="rgba(0,0,0,0)"),
        showlegend=False
    ))
    fig.add_trace(go.Scatter(
        x=scen_dfs[0].index.values[:24], y=global_max,
        mode="lines", name="Overlap Max",
        line=dict(color="rgba(0,0,0,0)"),
        showlegend=False,
        fill="tonexty", 
        fillcolor=overlap_color
    ))

    fig.add_trace(go.Scatter(
        x=df.index.values[:24], y=df["0"],
        mode="lines", name="Hist. DAP DE/LU",
        line=dict(color="black"),
        showlegend=True
    ))

    fig.update_layout(
        xaxis_title="Stunden",
        yaxis_title="DAP DE/LU in €/MWh",
        showlegend=True,
        legend=dict(
            x=0.01,
            y=0.99,
            bgcolor="rgba(255, 255, 255, 0.7)", 
            bordercolor="black",
            borderwidth=1
        )
    )
    fig.show()
```

```python
import scipy.stats as stats

def plot_dist(df, scen_dfs, bins=50):
    colors = px.colors.qualitative.Set1 
    fig = go.Figure()

    def histogram_edges(data, bins):
        counts, bin_edges = np.histogram(data, bins=bins)
        bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2  
        return bin_centers, counts, bin_edges

    bin_centers, counts, bin_edges = histogram_edges(df["0"], bins)
    fig.add_trace(
        go.Scatter(
            x=bin_centers, y=counts,
            mode="lines", name="Hist. DAP DE/LU",
            line=dict(color="black", width=2)
        )
    )

    for i, scen_df in enumerate(scen_dfs):
        bin_centers, counts, _ = histogram_edges(scen_df[f"Scenario_{i+1}"], bins)
        fig.add_trace(
            go.Scatter(
                x=bin_centers, y=counts,
                mode="lines", name=f"Szenarien",
                line=dict(color=colors[1], width=2) #colors=colors[i % len(colors)]
            )
        )

    fig.update_layout(
        xaxis_title="DAP DE/LU in €/MWh",
        yaxis_title="Anzahl",
        yaxis_type="log",
        yaxis=dict(tickvals=[10**i for i in range(-2, 5)]),
        xaxis=dict(tickmode="linear", dtick=100),
        showlegend=True,
        legend=dict(
            x=0.65,
            y=0.98,
            bgcolor="rgba(255, 255, 255, 0.7)", 
            bordercolor="black",
            borderwidth=1
        )
    )
    fig.show()
```
