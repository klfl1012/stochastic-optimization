```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_parquet("../scenario1.parquet")
```

```python
data.columns
```

### Feature Engineering

```python
data = data.resample("h", on="valid_time").mean().reset_index()
```

```python
# lag features of dap 1h, 2h, 3h
data["DA_Price_lag_1h"] = data["DA_Price"].shift(1)
data["DA_Price_lag_2h"] = data["DA_Price"].shift(2)
data["DA_Price_lag_3h"] = data["DA_Price"].shift(3)
```

```python
data.dropna(inplace=True)
```

```python
data.rename(columns={'valid_time': 'datetime'}, inplace=True)
```

```python
policy_test_data = data[data["datetime"] >= "2023-05-19"]
```

```python
policy_test_data.to_parquet("policy_test_data.parquet")
```
